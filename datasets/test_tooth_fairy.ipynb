```python
import sys
import os
import time
sys.path.append(os.getcwd().removesuffix('\\datasets'))
print(sys.path)
from torch.utils.data import DataLoader
import torch.nn.functional as F
from agh_class_segmentor import AGH_Class_Segmentor_Dataset
from data_prepare.utils_plot import plot_3d_fast
import numpy as np

dataset = AGH_Class_Segmentor_Dataset()
dataloader = DataLoader(dataset, batch_size=1)
for image, mask in dataloader:
    print(np.unique(mask.numpy()))
    num = np.unique(mask.numpy())[1]
    clean_output = F.one_hot(mask.long(), num_classes=34)  #32 classes
    clean_output = clean_output.permute(0, 4, 1, 2, 3).float()

    print(mask.max(), mask.min())
    print(clean_output.max(), clean_output.min())
    print(image.max(), image.min())
    plot_3d_fast(image[0, 0].numpy(), threshold=0.9)
    time.sleep(7)
    print(clean_output[0, num].max())
    plot_3d_fast(clean_output[0, num].numpy(), threshold=0.9)
    break
```
